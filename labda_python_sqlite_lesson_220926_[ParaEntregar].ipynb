```python
#1
import sqlite3
import pandas as pd

database_name = "empresa__zika_memo_company_3.db"
conn = sqlite3.connect(database_name, timeout=10.0)
cursor = conn.cursor()

table_name_employee = "Employees"
table_name_department = "Departments"

create_table_employee = f"""
CREATE TABLE IF NOT EXISTS {table_name_employee} (
  id INTEGER  PRIMARY KEY,
  name        TEXT,
  position    TEXT,
  income      REAL
);
"""

def check_connection(conn):
  try:
    conn.cursor()
    return 'Online' if True else 'Offline'
  except Exception as ex:
    return 'Offline'

print("Database Name:", database_name)
print("Connection Status:", check_connection(conn))
print("")

cursor.execute(create_table_employee)

cursor.execute(f"PRAGMA table_info({table_name_employee})")
columns = cursor.fetchall()

for i in columns:
  print("Column Name:", i[1].upper(), "- Column Type:", i[2])

```

#Parte 2: Manipulação de Dados (DML)

Inserção de dados (INSERT): Insira pelo menos 5 registros na tabela funcionários com dados fictícios.

Consulta de dados (SELECT): Escreva um comando SELECT para exibir todos os registros da tabela funcionários.

Filtro com WHERE:

Selecione apenas os funcionários com salário superior a R$ 5.000. Selecione os funcionários que trabalham como 'Gerente'.

Atualização de dados (UPDATE): Atualizar o salário de um funcionário específico. Por exemplo, aumentar o salário de ‘João’ em 10%.

Exclusão de dados (DELETE): Exclui um dos funcionários que você inseriu.

Contagem de registros: Conte o número total de funcionários na tabela.

```python
#Insert

excel_employees = {
    "name": ["Julia Vilela", "Salcy Lima", "Aline Midlej", "Marina Demori", "Gabriela Dias"],
    "position": ["Reporter", "Jornalist", "Presenter", "Director", "Presenter"],
    "income": [35000.00, 1000.00, 89000.90, 89900.00, 4000.00]
}

df = pd.DataFrame(excel_employees)

for _, row in df.iterrows():
  cursor.execute(f"""
    INSERT INTO {table_name_employee}
      (name, position, income)
    VALUES (?, ?, ?)
  """, (row["name"], row["position"], row["income"]))

conn.commit()

cursor.execute(f"""
                SELECT
                  *
                FROM {table_name_employee};
              """)
display(pd.DataFrame(cursor.fetchall(), columns = ["id", "name", "position", "income"]))
print("--")

cursor.execute(f"""
                SELECT
                  *
                FROM {table_name_employee}
                WHERE
                  income > 5000;
              """)

display(pd.DataFrame(cursor.fetchall(), columns = ["id", "name", "position", "income"]))
print("--")

cursor.execute(f"""
              UPDATE {table_name_employee}
              SET income = income + (income * 0.1)
              WHERE name = 'Salcy Lima'
              """)
conn.commit()

cursor.execute(f"""
                DELETE FROM {table_name_employee}
                WHERE name = 'Julia Vilela';
              """)
conn.commit()

cursor.execute(f"""
                SELECT
                  COUNT(*)
                FROM {table_name_employee}""")

print("TOTAL Employees: ", cursor.fetchone())
```

##Parte 3: Agregação e Ordenação


Funções
de agregação ( SUM, AVG, MAX, MIN):

Calcule a soma total dos salários.

Calcule o salário médio da empresa.

Encontre o salário mais alto e o mais baixo.

Agrupamento de dados ( GROUP BY):

Calcule um salário médio para cada cargo.

Conte quantos funcionários existem em cada carga.

Ordenação
( ORDER BY):

Selecione todos os funcionários e ordens de maior para o menor salário.

Selecione todos os funcionários e ordene-os por nome em ordem alfabética.

```python
#3

cursor.execute(f"""
                SELECT
                  SUM(income)
                  FROM {table_name_employee}""")
print("Total Income:", cursor.fetchone())

cursor.execute(f"""
                SELECT
                  AVG(income)
                  FROM {table_name_employee}""")
print("Average Income:", cursor.fetchone())

cursor.execute(f"""
                SELECT
                  MAX(income)
                  FROM {table_name_employee}""")
print("Bigger Income:", cursor.fetchone())

cursor.execute(f"""
                SELECT
                  MIN(income)
                  FROM {table_name_employee}""")
print("Smaller Income:", cursor.fetchone())

cursor.execute(f"""
                SELECT
                  position,
                  AVG(income)
                  FROM {table_name_employee}
                  GROUP BY position
                  ORDER BY position
               """)
display(pd.DataFrame(cursor.fetchall(), columns = ["position", "income"]))

cursor.execute(f"""
                SELECT
                  position,
                  COUNT(income)
                  FROM {table_name_employee}
                  GROUP BY position
                  ORDER BY position
               """)
display(pd.DataFrame(cursor.fetchall(), columns = ["position", "income"]))

cursor.execute(f"""
                SELECT
                  *
                  FROM {table_name_employee}
                  ORDER BY income DESC
               """)
display(pd.DataFrame(cursor.fetchall(), columns = ["id", "name", "position", "income"]))

select_all_employees = f"""
                SELECT
                  *
                  FROM {table_name_employee}
                  ORDER BY name
               """
cursor.execute(select_all_employees)
display(pd.DataFrame(cursor.fetchall(), columns = ["id", "name", "position", "income"]))
```

```python
#4
try:
    conn.execute("BEGIN;")

    cursor.execute(f"""
      INSERT INTO {table_name_employee}
        (
          name,
          position,
          income
        )
        VALUES
        (
          'Indianara Campos',
          'Reporter'
          20000
        )
    """)

    conn.commit()
except Exception as e:
    conn.rollback()
    print("Error! The command did not run succesfully. Detail:", e)
```
